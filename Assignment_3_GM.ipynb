##**2 - Analytical questions**
Let's create our questions. I have decided to look at a specific angle to develop my 6 questions: Gender behavior in micro-loans. It will guide us, throught the analysis of the dataset.

But first one problematic, we need to identify the gender for each loan,  

Here is the different problem to get there:
- 'female' contain 'male'
- there can be different gender for one loan
- Some gender are not specified (NaN)

So, to resolve that we need to detached each gender and create the gender_group column that will be either: wom, man, mixed, unknown.

```python
def gender(x):
  if pd.isna(x):
     return "unknown"
  y=set(x.split(", "))
  if y=={"female"}:
    return "wom"
  elif y=={"male"}:
    return "man"
  else:
    return "mixed"

loans["gender_group"] = loans["borrower_genders"].apply(lambda x: gender(x))
loans.head(15)
```

Now that it is done !

We will start by an overall thinking of the question, looking specificaly at the amount of loan per gender and the average loan amount for each.

***Question 1 : How many loans go to women, men, and mixed groups, and what is the average loan amount for each?***

```python
loans.groupby("gender_group")["loan_amount"].agg(["count", "mean"]).sort_values(by="count", ascending=False)
```

The difference in number of loans between men and women is considerable, with almost 3/4 of loans going to women. The average amount is different, slightly higher for men than women, but there is also a big gap when we look at mixed groups, probably because these loans are shared by several borrowers.

In another analysis, we could also check if the number of borrowers per loan affects the amount, as it may impact the results here.

Now, let's see more in details what is this representation by checking the repartition within the different sectors.

***Question 2 : In which sectors are women most (and least) represented among borrowers?***

```python
loans["is_woman"] = loans["gender_group"] == "wom"
loans.groupby('sector').agg(pct_women=('is_woman', 'mean'), avg_loan=('loan_amount', 'mean'), n_loans=('loan_amount', 'size')).round(2).sort_values(by='pct_women', ascending=False)
```

Women are the majority of borrowers in almost every sector. There is no clear link with the average amount: Retail and Food are very feminised with small loans, but Clothing and Arts are too, with loans above $1,000.

Now lets ask ourselves:

***Question 3 : Are loans to women funded faster than loans to men?***

```python
loans['posted_time'] = pd.to_datetime(loans['posted_time'])
loans['funded_time'] = pd.to_datetime(loans['funded_time'])
loans['days_to_fund'] = (loans['funded_time'] - loans['posted_time']).dt.days

loans[loans['funded_time'].notna()].groupby('gender_group').agg(mean_days=('days_to_fund', 'mean'), median_days=('days_to_fund', 'median'), n_loans=('days_to_fund', 'size')).round(1).sort_values(by='mean_days')
```

Loans to women are funded much faster. The mean is always higher than the median, which shows that a few very slow loans pull the average up, so the median is the better measure here.
We can also compare the column n_loans with the answer of Q1, and we will get the share of funded loans. That brings us to our 4th question:

***Question 4 : Which sectors have the highest share of loans that were not fully funded, and is it different for women and men?***

```python
loans['not_funded'] = loans['funded_amount'] < loans['loan_amount']
loans[loans['gender_group'].isin(['wom', 'man'])].groupby(['sector', 'gender_group']).agg(pct_not_funded=('not_funded', 'mean'), n_loans=('not_funded', 'size')).round(2).sort_values(by=['sector', 'pct_not_funded'], ascending=[True, False])
```

In every sector, loans to men are much more often left not fully funded than loans to women, usually 3 to 4 times more. Manufacturing and Wholesale are the only sectors where almost all loans get fully funded for both groups (but Wholesale has very few loans).

Let's check into one sector specifically:

***Question 5 : Among loans meant for education, what share go to women, and how does their average amount compare with all loans?***

```python
edu = loans[loans['use'].str.contains('school|tuition|education', case=False, na=False)]

edu.groupby('gender_group').agg(n_loans=('loan_amount', 'size'), avg_loan=('loan_amount', 'mean')).round(0).sort_values(by='n_loans', ascending=False)
```

Education loans are less feminised than loans in general , and men are more present. The gap in average amount almost disappears.

For the last question, let's look closely to the country repartition.

***Question 6 : In the five countries with the most loans, what share of large loans (above $1,000) go to women?***

```python
top5 = loans['country'].value_counts().head(5).index
large = loans[(loans['country'].isin(top5)) & (loans['loan_amount'] > 1000)]

large.groupby('country').agg(pct_women=('is_woman', 'mean'), n_loans=('is_woman', 'size')).round(2).sort_values(by='pct_women', ascending=False)
```

Among large loans in the five countries with the most loans, the share going to women varies a lot: from 84% in the Philippines to only 25% in Kenya.
